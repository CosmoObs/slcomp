```python
# result.to_parquet('delve_dr2_pixels.parquet')
result = pd.read_parquet("delve_dr2_pixels.parquet")
result.head()
```

```python
pixels_healpy = [
    hp.pix2ang(256, ipix=i, lonlat=True) for i in range(hp.nside2npix(256))
]
```

```python
pixels_healpy_df = pd.DataFrame(pixels_healpy, columns=["ra", "dec"])
pixels_healpy_df["ring256"] = pixels_healpy_df.index
```

```python
delve_pixels = pd.merge(pixels_healpy_df, result, on="ring256", how="inner")
len(delve_pixels), len(result)
```

```python
delve_pixels.to_csv("delve_coordinates.csv", index=False)
```
